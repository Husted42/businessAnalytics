```python
# Import packages for plotting
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

def catch(effort):
    return 1000 * effort - 0.4 * effort ** 2

# Get the MSY
def get_msy():
    effort = np.linspace(0, 2500, 100)
    catches = catch(effort)
    msy_index = np.argmax(catches)
    return effort[msy_index], catches[msy_index]

# Plot the catch-effort curve
def plot_catch_effort():
    effort = np.linspace(0, 2500, 100)
    catches = catch(effort)

    msy_effort, msy_catch = get_msy()

    plt.figure(figsize=(10, 6))
    sns.lineplot(x=effort, y=catches, label='Catch-Effort Curve', color='blue')
    plt.axvline(x=msy_effort, color='red', linestyle='--', label='MSY Effort')
    plt.axhline(y=msy_catch, color='green', linestyle='--', label='MSY Catch')
    plt.scatter([msy_effort], [msy_catch], color='black')

    plt.text(msy_effort + 50, msy_catch - 50, f'MSY: ({msy_effort:.1f}, {msy_catch:.1f})', color='black')
    plt.title('Catch-Effort Curve with MSY')
    plt.xlabel('Effort')
    plt.ylabel('Catch')
    plt.legend()
    plt.grid()
    plt.show()

if __name__ == "__main__":
    plot_catch_effort()
```
