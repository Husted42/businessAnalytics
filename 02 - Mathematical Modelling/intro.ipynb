```julia
using JuMP, HiGHS

# Create a model
model = Model(HiGHS.Optimizer)

# Define the variables
@variable(model, x >= 0)
@variable(model, y >= 0)

# Add constraint
@constraint(model, x + y <= 1)

# Set objective
@objective(model, Max, x + 2y)

# Optimize
optimize!(model)

# Print results
println("Optimal solution:")
println("x = ", value(x))
println("y = ", value(y))
```

The jerwlery problem

```julia
using JuMP, HiGHS
full_day = 7.5 * 60 # in minutes

# Create a model
model = Model(HiGHS.Optimizer)

# Define the variables
@variable(model, x_1 >= 0)
@variable(model, x_2 >= 0)
@variable(model, x_3 >= 0)
@variable(model, x_4 >= 0)
@variable(model, x_5 >= 0)

# Machines
@constraint(model, 7 * x_1 + 9 * x_4 <= full_day) # Machine 1
@constraint(model, 5 * x_1 + 7 * x_2 + 11 * x_3 + 5 * x_5 <= full_day) # Machine 2
@constraint(model, 3 * x_2 + 8 * x_3 + 15 * x_4 + 3 * x_5 <= full_day) # Machine 3

# Assembly
@constraint(model, 12 * x_1 + 3 * x_2 + 11 * x_3 + 9 * x_4 + 6 * x_5 <= full_day*2)

# Defining the objective function
@objective(model, Max, 50 * x_1 + 45 * x_2 + 85 * x_3 + 60 * x_4 + 55 * x_5)

# Optimize
optimize!(model)

if termination_status(model) == MOI.OPTIMAL
    println("Optimal objective value: $(objective_value(model))")
else
    println("No optimal solution found.")
end
```

### Excercice 1.2 - The Jewlery problem

```julia
using JuMP, HiGHS
full_day = 7.5 * 60 # in minutes

function jewrlery(workers::Int)
    

    # Create a model
    model = Model(HiGHS.Optimizer)
    set_optimizer_attribute(model, "log_to_console", false)

    # Define the variables
    @variable(model, x_1 >= 0)
    @variable(model, x_2 >= 0)
    @variable(model, x_3 >= 0)
    @variable(model, x_4 >= 0)
    @variable(model, x_5 >= 0)

    # Machines
    @constraint(model, 7 * x_1 + 9 * x_4 <= full_day) # Machine 1
    @constraint(model, 5 * x_1 + 7 * x_2 + 11 * x_3 + 5 * x_5 <= full_day) # Machine 2
    @constraint(model, 3 * x_2 + 8 * x_3 + 15 * x_4 + 3 * x_5 <= full_day) # Machine 3

    # Assembly
    @constraint(model, 12 * x_1 + 3 * x_2 + 11 * x_3 + 9 * x_4 + 6 * x_5 <= full_day*workers)

    # Demand
    @constraint(model, x_1 <= 25)
    @constraint(model, x_2 <= 10)
    @constraint(model, x_3 <= 12)
    @constraint(model, x_4 <= 15)
    @constraint(model, x_5 <= 60)
    # Defining the objective function
    @objective(model, Max, 50 * x_1 + 45 * x_2 + 85 * x_3 + 60 * x_4 + 55 * x_5)

    # Optimize
    optimize!(model)

    if termination_status(model) == MOI.OPTIMAL
        println("Optimal objective value: $(objective_value(model))")
    else
        println("No optimal solution found.")
    end


end

for i in 1:5
    println("With $i workers:")
    jewrlery(i)
    println()
end
```

### Excercice 1.3 - The brewery problem

```julia
using JuMP, HiGHS

# Create a model
model = Model(HiGHS.Optimizer)
set_optimizer_attribute(model, "log_to_console", true)

months = 1:12
demand = [15, 30, 25, 55, 75, 115, 190, 210, 105, 65, 20, 20]
production_capacity = [120 for m in months]


# Production variables
@variable(model, p[months] >= 0) # Production
@variable(model, s[months] >= 0) # Storage

@constraint(model, p .<= production_capacity)

# Inventory balance constraints
@constraint(model, [m in months],
    s[m] == (m > 1 ? s[m-1] : 0) + p[m] - demand[m]
)

@objective(model, Min, sum(s[m] for m in months))

optimize!(model)

if termination_status(model) == MOI.OPTIMAL
    println("Optimal objective value: $(objective_value(model))")
    println("Workers: $(value(w_0))")
else
    println("No optimal solution found.")
end
```
