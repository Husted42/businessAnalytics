Consider the entropy function
$$
    J(y) = - \int^1_0 y(x) log(y(x)) dx
$$ 

Find the directional derivative $dJ(Y)(\eta)$ of J at y given $y(x) = x + \frac{1}{2}$


```python
import numpy as np
import matplotlib.pyplot as plt
# x-space
dx = 0.01
x = np.arange(0, 1, dx)

def y(x):
    return x + 1/2

def dfdy(x):
    return -1-np.log(y(x))

def eta(x):
    return 1/2 - x

plt.plot(x, y(x), label='y(x)')
plt.plot(x, dfdy(x), label="dy/dx")
plt.ylim(-2, 2)
plt.xlim(0, 1)
plt.xlabel('x')
plt.ylabel('y, dy/dx')
plt.show()
```

Riemann sum approximation

$ \int^1_0 f(x) dx \approx \sum^{N-1}_{i=0} f(x_i) \:\: \Delta x $

```python
dj = np.sum(dfdy(x)*eta(x)) * dx
print("Integral of dy/dx * eta(x) over [0,1]:", dj)
```

To increase the entropy, how should probability be redistribute? We see as the probability x increasses the entropy increases as well.
