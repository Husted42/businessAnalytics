```python
import requests

api_key = "<GETYOURAPIKEY>"
url_base = "https://app.ticketmaster.com/discovery/v2/events.json"
params = ["countryCode=DK", "apikey="+api_key, "size=200","startDateTime="+datetime.datetime.now().strftime("%Y-%m-%dT%H:%M:%SZ")]

request_url = url_base + "?" + "&".join(params)

response = requests.get(request_url)
data = response.json()
```

```python
data.keys()
```

```python
for e in data['_embedded']['events']:
    if e['classifications'][-1]['segment']['name']=="Music":
        print(e['name'], e['dates']['start']['localDate'], "\t", end =" ")
        for t in e['classifications']:
            print(t['subGenre']['name'], end =" ")
        print("")
```
