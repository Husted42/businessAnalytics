In this notebook, we explore the EEA's dataset API. In their system, they assume that each call is essentially an SQL query. This implies that we need to know well the databases that they have.

#### 1.1. A simple example

First, the imports...

```python
import requests
import json
import matplotlib.pyplot as plt
import pandas as pd
```

We will not be able to teach you about SQL. For what you need for this, the amount of SQL required is actually pretty basic (just google for the sintax of the SQL select command). 

Here is an example, where we want to get the table "metada" that is in the V2 version of the "metadata" database:

```python
http="https://discodata.eea.europa.eu/sql?query=Select * from [metadata].[v2].[metadata]"
res = requests.get(http)
```

```python
len(res.content)
```

This is actually a LOT of data, so unless you are really patient (and have a good computer), we recommend you to pick specific fields. Here's an example

```python
jsonres=json.loads(res.content)
```

```python
eea_m_df=pd.DataFrame(jsonres['results'])
```

```python
eea_m_df
```

#### 1.2. Biodiversity data

We will start with a table with generic statistics about the different countries. Just take a look

```python
url="https://discodata.eea.europa.eu/sql?query=Select * from [EUNIS].[v1r3].[Art12_mastertab_data_detail]"
```

```python
jsonres
```
