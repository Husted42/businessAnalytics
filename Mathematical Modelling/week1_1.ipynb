```julia

using JuMP, HiGHS

months = 1:12



# Create a model
model = Model(HiGHS.Optimizer)
set_optimizer_attribute(model, "log_to_console", true)


@variable(model, v_1 >= 0)
@variable(model, v_2 >= 0)
@variable(model, o_1 >= 0)
@variable(model, o_2 >= 0)
@variable(model, o_3 >= 0)

@objective(model, Max, 40 * v_1 + 30 * v_2 + 20 * o_1 + 40 * o_2 + 35 * o_3)

@constraint(model, v_1 + v_2 <= 200)
@constraint(model, o_1 + o_2 + o_3 <= 250)

@constraint(model, 
    (8.8 * v_1 +
    6.1 * v_2 +
    2.0 * o_1 +
    4.2 * o_2 +
    5.0 * o_3)  <= 6 * (v_1 + v_2 + o_1 + o_2 + o_3)
)

@constraint(model, 
    (8.8 * v_1 +
    6.1 * v_2 +
    2.0 * o_1 +
    4.2 * o_2 +
    5.0 * o_3)  >= 3 * (v_1 + v_2 + o_1 + o_2 + o_3)
)

optimize!(model)

println("Optimal solution:")
println("z = ", (objective_value(model)))
println("v_1 = ", value(v_1))
println("v_2 = ", value(v_2))
println("o_1 = ", value(o_1))
println("o_2 = ", value(o_2))
println("o_3 = ", value(o_3))

```
